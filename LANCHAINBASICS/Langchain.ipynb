```python
## dynamic prompt templates
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableSequence, RunnableMap, RunnablePassthrough,RunnableLambda

## create a translation prompt template
translation_template = ChatPromptTemplate.from_messages([
    ("system", "You are a profeesional translator . Translate the following {text} from {source_language} to {target_language} and also maintain the tone and style of the text."),
    ("user", "{text}")
])

## using the template
prompt = translation_template.invoke({
    "source_language":"English",
    "target_language":"French",
    "text":"Hello, how are you?"
 })
```

```python
prompt
```

```python
translated_response = model.invoke(prompt)
```

```python
print(translated_response.content)
```

```python
##BUIDLING YOUR FIRST CHAIN 

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


def create_story_chain():

    # Prompt for story generation
    story_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a professional story writer. Write a short story based on the following theme, character, and setting."),
        ("user", "Theme: {theme}\nMain Character: {characters}\nSetting: {setting}")
    ])

    # Prompt for story analysis
    analysis_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a professional story analyst. Analyze the following story and provide a brief summary, key themes, and character analysis."),
        ("user", "{story}")
    ])

    # Story generation chain
    story_chain = story_prompt | model | StrOutputParser()
    # create a function to pass proper insights

    def analyse_story(story_text):
        return {"story": story_text}
    # Story analysis chain
    analysis_chain = (
        story_chain
        | RunnableLambda(analyse_story)
        | analysis_prompt
        | model
        | StrOutputParser()
    )

    return analysis_chain
```

```python
chain = create_story_chain()
chain
```

```python
chain = create_story_chain()
chain
```

```python
result = chain.invoke({
    "theme": "Friendship and Adventure",
    "characters": "Alice, Bob, and Charlie",
    "setting": "A magical forest"
})

print("STORY AND ANALYSIS:")
print(result)
```
