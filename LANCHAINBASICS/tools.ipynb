```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """
    it will return the sum of two numbers a and b

    args: 
        a: Number One
        b: Number Two
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int):
    """
    it will return the product of two numbers a and b

    args: 
        a: Number One
        b: Number Two
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
agent = create_agent(
    model =llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt = "You are a helpful assistant that can perform addition and multiplication of two numbers. " 
)
```

```python
response = agent.invoke({"messages":[{"role":"user", "content":"Please add 5 and 3 *100"} ]})
```
