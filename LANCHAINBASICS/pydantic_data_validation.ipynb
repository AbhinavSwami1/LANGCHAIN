```python
from pydantic import BaseModel,Field
from typing import List,Annotated
```

```python
class User(BaseModel):
    name:str
    age:int
```

```python
user=User(name="abhinav", age=20)
```

```python
user
```

```python
### FIELD VALIDATION
```

```python
class Product(BaseModel):
    title:str =Field(max_length=6)
    des: str= Field(max_length=200)
    price : int= Field(gt=0, le=100)
```

```python
product= Product(title="samsung234", des="a best phone", price=20000)
```

```python
product= Product(title="xiamo", des="a best phone", price=20)
```

```python
class Order(BaseModel):
    id :str= Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id="rmnv7343jb",product="bjhsdc")
```

```python
order = Order(id="rmnv7343jb",product=product)
```

```python
order
```

```python
order.product
order.product.title
```

```python
class Carts(BaseModel):
    id :str= Field(min_length=6, max_length=12)
    items : List[Product]
    comment :str=Field(default="No Comments")
```

```python
cart= Carts(id="2131d2ec", items=product)
```

```python
cart= Carts(id="2131d2ec", items=[product])
```

```python
cart
```

```python
for product in cart.items:
    print(product)
```

```python
## annotator

common_Validator=Annotated[List[Product], "This is Common Validator"]
```

```python
class Carts(BaseModel):
    id :str= Field(min_length=6, max_length=12)
    items : common_Validator
    comment :str=Field(default="No Comments")
```

```python
class Order(BaseModel):
    products: common_Validator
    
```

```python
ord = Order(products=[product])
```
